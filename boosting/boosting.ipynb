```python
from xgboost import XGBClassifier,XGBRegressor
from sklearn.model_selection import train_test_split
import pandas as pd
from sklearn.metrics import f1_score, classification_report


```

```python
%pip install xgboost
```

```python
df=pd.read_csv('loan_data.csv')
df.head()
```

```python
x=df.drop(columns='loan_status')
y=df['loan_status']
cat_cols=x.select_dtypes(include='object').columns
cat_cols
x[cat_cols]=x[cat_cols].astype('category')
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
model=XGBClassifier (
    n_estimators=100,
    max_depth=None,
    objective='binary:logistic', #multi:softprob 
    enable_categorical=True

)
model.fit(xtrain,ytrain)
```

##Advantages
1.Scaling and encoding techniquest are not required

it handle missing values


```python
ytrain_predict=model.predict(xtrain)
xtrain_predict=model.predict(xtest)
```

```python
model_eval=f1_score(ytrain,ytrain_predict)
model_eval
```

```python
report=classification_report(ytrain,ytrain_predict)
print(report)
```

```python
report2=classification_report(ytest,xtrain_predict)
print(report2)
```
