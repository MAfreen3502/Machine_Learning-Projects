```python
X=df[['Temperature']]
Y=df[['Revenue']]
```

```python
print(X.head)
```

```python
X_tain,X_test,y_train,y_test=train_test_split(X,Y,train_size=0.8,random_state=42)
```

```python
print(X_tain.shape)
print(X_test.shape)
```

```python
model=LinearRegression()
```

```python
model.fit(X_tain,y_train)
```

```python
ytrain_pred=model.predict(X_tain)
```

```python
print("Actual Revenue:")
print(y_test.head())

```

```python
print("Predicted Revenue:")
print(ytrain_pred[:5])
```

```python
mae = mean_absolute_error(y_train, ytrain_pred)

print("MAE:", mae)
```

```python
mse = mean_squared_error(y_train, ytrain_pred)

print("MSE:", mse)
```

```python
rmse = np.sqrt(mse)

print("RMSE:", rmse)
```

```python
r2 = r2_score(y_train, ytrain_pred)

print("R2 Score:", r2)
```

```python
ytest_pred=model.predict(X_test)
r1=r2_score(y_test,ytest_pred)
print(r1)
```

# Conclusion #
The Linear Regression model was trained using Temperature as the input feature and Revenue as the target variable.

The model achieved an R² Score of approximately 0.98034 on the training data,

 meaning it explains about 98% of the variation in Revenue in that data. 
 
 This indicates a strong linear relationship captured by the model.
