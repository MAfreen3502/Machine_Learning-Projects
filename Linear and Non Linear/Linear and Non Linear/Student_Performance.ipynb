```python


fig, axes = plt.subplots(2, 3, figsize=(15, 8))  # 2 rows, 3 columns = 6 plots

pairs = [
    ('study_hours', 'math_score'),
    ('study_hours', 'science_score'),
    ('study_hours', 'english_score'),
    ('study_hours', 'overall_score'),
    ('attendance_percentage', 'overall_score'),
    ('math_score', 'overall_score')
]

for ax, (x, y) in zip(axes.flat, pairs):
    ax.scatter(sp_clean[x], sp_clean[y], alpha=0.4, s=15)
    ax.set_xlabel(x)
    ax.set_ylabel(y)
    ax.set_title(f'{x} vs {y}')

plt.tight_layout()
plt.show()
```
