```python
from sklearn.compose import ColumnTransformer

preprocessing=ColumnTransformer(
    transformers=[
        ('order_encoder',OrdinalEncoder(
            categories=[['High School','Associate','Master','Doctarate']],
            handle_unknown='use_encoded_value',
            unknown_value=-1),['person_education']),
            ('onehot_encoder',OneHotEncoder(sparse_output=False,handle_unknown='error'),
            ['person_gender','person_home_ownership','loan_intent','previous_loan_defaults_on_file'])
    ],remainder="drop"
)
preprocessing.fit_transform(xtrain)
```
