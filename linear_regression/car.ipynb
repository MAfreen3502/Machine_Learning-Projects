```python
sns.scatterplot(x=df["Volume"], y=df["CO2"])
plt.xlabel("Volume")
plt.ylabel("CO2")
plt.title("Volume vs CO2")
plt.show()
```

## Volume vs CO₂ → 0.592082

Since the correlation is positive (0.592), Volume and CO₂ have a moderate positive linear relationship.

Conclusion: There is a moderate positive relationship between Volume and CO₂ emissions. 
As the engine Volume increases, CO₂ emissions tend to increase.

```python
sns.scatterplot(x=df["Weight"], y=df["CO2"])
plt.xlabel("Weight")
plt.ylabel("CO2")
plt.title("Weight vs CO2")
plt.show()
```

## Weight vs CO₂ → 0.552150

Since the correlation is positive (0.552), Weight and CO₂ also have a moderate positive linear relationship.

Conclusion: There is a moderate positive relationship between Weight and CO₂ emissions.
 As the weight of the car increases, CO₂ emissions tend to increase.

```python
sns.scatterplot(x=df["Volume"], y=df["Weight"])
plt.xlabel("Volume")
plt.ylabel("Weight")
plt.title("Volume vs Weight")
plt.show()
```

## Volume vs Weight → 0.753537

This is the strongest correlation among the three pairs.

Conclusion: There is a strong positive relationship between Volume and Weight. 
As the engine Volume increases, the weight of the car tends to increase.
