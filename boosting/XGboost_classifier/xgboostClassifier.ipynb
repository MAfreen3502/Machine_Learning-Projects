```python
from xgboost import XGBClassifier
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, classification_report
```

```python
df=pd.read_csv('vitamin_deficiency_disease_dataset_20260123.csv')
df.head()
```

```python
x=df.drop(columns='has_multiple_deficiencies')
y=df['has_multiple_deficiencies']
```

```python
x
```

```python
cat_cols=x.select_dtypes(include='object').columns
cat_cols
x[cat_cols]=x[cat_cols].astype('category')
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
model=XGBClassifier (
    n_estimators=100,
    max_depth=None,
    objective='binary:logistic', #multi:softprob 
    enable_categorical=True

)
model.fit(xtrain,ytrain)
```

here we using xtest for predict ,not ytest bcz xtest consist of questions

in model evaluation we verify how well that match to actual ytest answers

```python
y_pred=model.predict(xtest)
xtrain_predict=model.predict(xtrain)
```

```python
f1_score1=f1_score(ytest,y_pred)
f1_score1
```

```python
xtrain_score=f1_score(ytrain,xtrain_predict)
xtrain_score
```

```python
report1=classification_report(ytrain,xtrain_predict)
print(report1)
```

```python
report=classification_report(ytest,y_pred)
print(report)
```

```python
from sklearn.metrics import accuracy_score

print("Training Accuracy:",
      accuracy_score(ytrain, xtrain_predict))

print("Testing Accuracy:",
      accuracy_score(ytest, y_pred))
```
